### X Import Libraries
First, we'll import the necessary libraries for data manipulation and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

### X Load the Data
Load the `BOIL ETF Stock Price History.csv` file into a pandas DataFrame.

```python
df = pd.read_csv('/content/BOIL ETF Stock Price History.csv')
display(df.head())
```

### X Initial Data Inspection
Let's get an overview of the dataset, check data types, and look for missing values.

```python
print(df.info())
print('\nDataFrame Description:')
display(df.describe())
```

### X Data Cleaning and Preprocessing

- Convert the 'Date' column to datetime objects.
- Convert relevant columns to numeric types, handling non-numeric values.
- Rename columns for easier access (e.g., remove spaces or special characters).

```python
# Convert 'Date' to datetime and set as index
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values(by='Date')
df = df.set_index('Date')

# Clean column names (remove special characters and strip whitespace)
df.columns = df.columns.str.replace('[^A-Za-z0-9_]+', '', regex=True).str.strip()

# Convert relevant columns to numeric, handling commas and percentages
for col in ['Price', 'Open', 'High', 'Low', 'Vol', 'Change']: # Assuming these are the numeric columns
    if col == 'Vol':
        # Remove 'M' for millions and convert to numeric
        df[col] = df[col].astype(str).str.replace('M', '', regex=False).astype(float) * 1_000_000
    elif col == 'Change':
        # Remove '%' and convert to numeric
        df[col] = df[col].astype(str).str.replace('%', '', regex=False).astype(float) / 100
    else:
        # Remove commas from price columns and convert to float
        df[col] = df[col].astype(str).str.replace(',', '', regex=False).astype(float)

display(df.head())
```

### X Feature Engineering (Optional)

Calculate daily returns and moving averages to enrich the dataset for further analysis.

```python
# Calculate daily returns
df['DailyReturn'] = df['Price'].pct_change()

# Calculate moving averages
df['MA_10'] = df['Price'].rolling(window=10).mean()
df['MA_50'] = df['Price'].rolling(window=50).mean()

display(df.tail())
```

### X Data Visualization

Visualize the stock price history, including closing prices and moving averages.

```python
plt.figure(figsize=(14, 7))
plt.plot(df.index, df['Price'], label='Closing Price', color='skyblue', alpha=0.7)
plt.plot(df.index, df['MA_10'], label='10-Day MA', linestyle='--', color='orange')
plt.plot(df.index, df['MA_50'], label='50-Day MA', linestyle='--', color='green')

# Dates to highlight
highlight_dates_str = []
highlight_dates = pd.to_datetime(highlight_dates_str)

# Filter DataFrame for these dates
highlight_df = df.loc[df.index.isin(highlight_dates)]

# Plot the highlighted dates as scatter points
plt.scatter(highlight_df.index, highlight_df['Price'], color='red', s=100, zorder=5, label='Highlighted Dates', marker='X')

plt.title('BOIL ETF Stock Price History with Moving Averages and Highlighted Dates')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.histplot(df['DailyReturn'].dropna(), bins=50, kde=True)
plt.title('Distribution of Daily Returns')
plt.xlabel('Daily Return')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```
