```python
def american_to_percentage_odds(american_odds):
    """
    Converts American odds to percentage probability.

    Args:
        american_odds (int): American odds (e.g., -110, +150).

    Returns:
        float: The implied probability in percentage (0-100).
    """
    if american_odds > 0:
        percentage_odds = 100 / (american_odds + 100)
    elif american_odds < 0:
        percentage_odds = abs(american_odds) / (abs(american_odds) + 100)
    else:
        return 0.5  # Even odds, X% chance, though usually represented as X or -X

    return percentage_odds * 100

# User-defined list of American odds inputs:
american_odds_list = [255, 250]

# Iterate through the list and print individual outputs
for odds in american_odds_list:
    percentage = american_to_percentage_odds(odds)
    sign = '+' if odds >= 0 else ''
    print(f"American odds of {sign}{odds} is equivalent to {percentage:.2f}% probability.")
```

```python
def percentage_to_american_odds(percentage_odds):
    """
    Converts percentage probability to American odds.

    Args:
        percentage_odds (float): The implied probability in percentage (0-100).

    Returns:
        int: American odds (e.g., -110, +150).
    """
    probability = percentage_odds / 100

    if probability >= 0.5:
        american_odds = -100 * probability / (1 - probability)
    else:
        american_odds = 100 * (1 - probability) / probability

    return round(american_odds)
```

```python
def calculate_initial_capital(american_odds, desired_total_return):
    """
    Calculates the initial capital needed to achieve a desired total return based on American odds.

    Args:
        american_odds (int): American odds (e.g., -110, +150).
        desired_total_return (float): The total amount to be received back if the bet wins (initial stake + profit).

    Returns:
        float: The initial capital required.
    """
    if american_odds > 0:
        # For positive odds (underdog), e.g., +X means $X stake wins $X profit, total return $X.
        # To get a desired_total_return, initial capital = desired_total_return * (X / (X + american_odds))
        initial_capital = desired_total_return * (100 / (100 + american_odds))
    elif american_odds < 0:
        # For negative odds (favorite), e.g., -X means $X stake wins $X profit, total return $X.
        # To get a desired_total_return, initial capital = desired_total_return * (abs(american_odds) / (X + abs(american_odds)))
        initial_capital = desired_total_return * (abs(american_odds) / (100 + abs(american_odds)))
    else:
        # Even odds (e.g., +X or -X usually), where stake equals profit.
        # If desired_total_return is Xx stake, then initial_capital = desired_total_return / X
        initial_capital = desired_total_return / 2 # Assuming X:X payout for even odds if total return is specified

    return initial_capital
```
